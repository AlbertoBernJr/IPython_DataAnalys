1. enumerate = Ao processar linhas de um arquivo CSV, logs ou API, é comum precisar do índice (ex: para tratar erros na linha 120).

2. sorted = Ordenar dados (ex: top 10 clientes por valor de compra, logs por data, KPIs por ranking).

3. zip = Combinar colunas vindas de diferentes fontes, ou criar dicionários que unem chaves e valores.  

4. reversed = Verificar últimas entradas, como as últimas transações, acessos ou leituras de sensor.   

5. List comprehensions = Criar colunas derivadas, limpar dados, extrair padrões de texto, transformar listas em datasets estruturados.

6. Set comprehensions = Identificar valores únicos, categorias distintas, IDs sem repetição. 

7. Dict comprehensions = Mapear IDs para nomes, criar lookup tables, indexar registros.

8. Nested comprehensions = “Achatar” listas aninhadas, extrair dados de estruturas complexas (como JSONs ou logs com múltiplas camadas).


```python
clientes = [
    {"nome": "Alice", "gasto": 2400},
    {"nome": "Bruno", "gasto": 500},
    {"nome": "Carlos", "gasto": 3200},
    {"nome": "Diana", "gasto": 1800},
]

#ordenar do maior para o menor
clientes2 = sorted(clientes, key=lambda x: x['gasto'], reverse=True)
"""
[key=lambda x: x['gasto']] -> usa o valor do campo [gasto] como criterio de ordenação
[reverse=True] -> ordena em ordem decrescente (maior para menor)
"""

#os 3 primeiros
top_3 =  clientes2[:3] #[ [:3] ] -> pega os 3 primeiros elementos da lista ordenada

print("Top 3:")
for clientes in top_3:
    print(f"{clientes['nome']}: R$ {clientes['gasto']}")
```

```python
"""
zip – Combinando colunas

Cenário:
Você recebeu duas listas separadas com dados de um mesmo dataset exportado do banco.
"""

nomes = ["Alice", "Bruno", "Carlos", "Diana"]
idades = [25, 32, 28, 41]

junto = dict(zip(nomes, idades))
#[zip(x, y)] = cria pares com os elementos -> (x, y)
#[dict()] = transforma os pares em dicionário
print(junto)

# Criando lista de dicionários
clientes = [{"nome": nome, "idades": idades} for nome, idades in zip(nomes, idades)]
print(clientes)
```

```python
"""
reversed – Últimos registros

Cenário:
Você precisa verificar os últimos 5 acessos de um log de sistema.
"""

acessos = ["user_1", "user_2", "user_3", "user_4", "user_5", "user_6", "user_7"]
ultimos = acessos[-3:] #Usando slicing

print(ultimos)
```

```python
"""
List comprehensions – Limpeza de dados

Cenário:
Você recebeu uma lista com possíveis nomes de produtos, com alguns valores nulos e espaços extras.
"""

produtos = ["  Maçã", None, "Banana ", "  ", "Laranja", ""]

produtos_limpos=[
    produto.strip() #remove espaços extras
    for produto in produtos #interage com cada produto
    if produto and produto.strip() #filtra não [None] e não vazio depois do [strip]
]

print("Produtos limpos:", produtos_limpos)
```

```python
"""
Set comprehensions – Identificando categorias únicas

Cenário:
Você extraiu uma lista de pedidos, e quer saber as categorias únicas de produtos.
"""

categorias = ["fruta", "bebida", "fruta", "limpeza", "bebida", "fruta"]
# Use set comprehension para gerar um conjunto de categorias únicas.

unico = {categoria for categoria in categorias}
print(unico)
```

```python
"""
Dict comprehensions – Criando mapas de lookup

Cenário:
Você precisa criar um dicionário de referência de IDs de produto para nome.
"""

ids = [101, 102, 103, 104]
nomes = ["Maçã", "Banana", "Laranja", "Uva"]

junto = dict(zip(ids, nomes))
print(junto)
```

```python
"""
comprehensions – Flattening de dados*

Cenário:
Você extraiu dados de uma API que retorna listas aninhadas de valores de sensor.
"""

leituras = [
    [23.1, 22.8, 23.0],
    [22.5, 22.7, 22.9],
    [23.3, 23.1, 23.4]
]

print("Lista em MAP:") #[MAP] = converte os elementos para string
for elemento in map(str, leituras):
    print(elemento)
    
print(" ") 
print("Lista numa unica linha:")
print(", ".join(map(str, leituras))) 
#[Join] so funciona com string
```
