Por que Sets são Essenciais?
1. Deduplicação Rápida: É a forma mais rápida em Python de encontrar ou contar valores distintos em uma coleção de dados.

2. Verificação de Existência (Lookups) O(1): Verificar se um item existe em um set (if item in meu_set:) é extremamente rápido (tempo constante, ou O(1)), mesmo que o set tenha milhões de itens. Fazer a mesma verificação em uma list (if item in minha_lista:) é muito lento (tempo linear, ou O(n)), pois o Python precisa verificar item por item.

3. Reconciliação de Dados: As operações de diferença (-) e interseção (&) são fundamentais para comparar dois conjuntos de dados (ex: dados da origem vs. dados do destino) e encontrar discrepâncias.

```python
"""
Contagem de Usuários Distintos

Conceitos: Criação de set a partir de uma lista para obter 
unicidade.

Cenário: Você tem um log de acessos ao site e seu gerente 
pergunta: "Quantos usuários distintos acessaram o sistema hoje?" 
Você não quer o total de cliques, mas sim o total de IDs únicos.
"""
# Lista de IDs de usuários a cada acesso (muitos repetidos)
log_acessos = [
    'user-123', 'user-456', 'user-123', 'user-789', 'user-456',
    'user-123', 'user-999', 'user-789', 'user-001', 'user-456'
]

print(f"Total de acessos no dia: {len(log_acessos)}")

#Converter a lista para um [set] e remover a duplicidade
usuarios_unicos=set(log_acessos)

print(f"Total de usuários distintos: {len(usuarios_unicos)}")
print(f"Ids dos usuários distintos: {usuarios_unicos}")
```

```python
"""
Reconciliação de Dados (Data Reconciliation)

Conceitos: Operações de diferença (- ou .difference()).

Cenário: Você executou um pipeline de ETL que deveria mover dados 
de um banco de dados de origem para um Data Warehouse (destino). 
Você precisa verificar se algum registro se perdeu no caminho.
"""

# IDs que você leu da tabela de ORIGEM
ids1 = {1001, 1002, 1003, 1004, 1005, 1006}

# IDs que foram gravados com sucesso na tabela de DESTINO
ids2 = {1001, 1003, 1004, 1005}

ids_falhas=ids1-ids2
print(f"IDs que estão em [1], mas faltam em [2]: {ids_falhas}")
```

```python
"""
Deduplicação de Registros Inteiros

Conceitos: Imutabilidade (listas -> tuplas) para usar em sets.

Cenário: Sua fonte de dados está "suja" e envia registros inteiros 
duplicados. Um registro é uma lista [id_pedido, id_cliente, valor]. 
Você quer uma coleção de registros únicos.

Como o texto explicou, você não pode colocar uma list dentro de 
um set (porque listas são mutáveis). A solução é convertê-las 
para tuple (que são imutáveis).
"""

dados_duplicados = [
    [1, 10, 50.0],
    [2, 20, 30.0],
    [1, 10, 50.0],  # Duplicado
    [3, 30, 75.0],
    [2, 20, 30.0]   # Duplicado
]

#armazenar as tuplas
registros_unicos = set()

for i in dados_duplicados:
    #converter a lista mutavel em tupla imutável
    registro_tupla=tuple(i)

    #adiciona a tupla ao [set]
    registros_unicos.add(registro_tupla)

print("Registros unicos - tupla:")
print(registros_unicos)

#de tupla para uma lista de novo
registros_lista=[list(tupla) for tupla in registros_unicos]
print("\nRegistros unicos - lista:")
print(registros_lista)
```

```python
"""
Segmentação de Clientes

Conceitos: interseção (& ou .intersection()).

Cenário: O time de Marketing quer uma lista de clientes "altamente 
engajados". A definição deles é: "clientes que compraram no último 
mês E também acessaram o aplicativo no último mês".

Sua Tarefa: Dadas duas listas de IDs, encontre os IDs que estão em
ambas.
"""

ids_compra = [101, 102, 103, 104, 105, 201, 202]
ids_app = [103, 104, 105, 106, 107, 202, 301]

#adicionando em [set] para retirar duplicatas
set_compra=set(ids_compra)
set_app=set(ids_app)

#comparando valores que tem nos 2 sets - usando [&]
comparacao=set_app&set_compra
print(comparacao)
```
